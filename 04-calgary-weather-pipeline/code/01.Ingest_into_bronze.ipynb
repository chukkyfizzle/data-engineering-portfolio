```python
from pyspark.sql.functions import col, year, month, quarter

table_name = 'bronze_table'

df = spark.read.format('json').load('Files/weather_data')
df.printSchema()
```

```python
bronze_table = "bronze_weather_data"

df = df \
.withColumnRenamed("Wind KPH", "Wind_KPH")
bronze_df = df.write \
.mode("append") \
.format("delta") \
.saveAsTable(bronze_table)
```
